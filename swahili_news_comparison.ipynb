# Model comparison

Accuracy, macro-F1, training time, and checkpoint size for the models that have been scored. Each file is written by that model's notebook. A missing file means that run has not been saved yet.

The test numbers are from the official test split, scored once after the checkpoint was chosen on validation macro-F1. Training time is the fit only, not data download.


```python
import json
from pathlib import Path

import pandas as pd

files = [
    ("TF-IDF + Logistic Regression", Path("baseline_metrics.json")),
    ("fastText cc.sw.300 + BiGRU", Path("bigru_metrics.json")),
    ("AfroXLMR-base", Path("afroxlmr_metrics.json")),
]
rows = []
for name, path in files:
    if not path.exists():
        rows.append({
            "model": name + " (not run yet)",
            "test_accuracy": None,
            "test_macro_f1": None,
            "train_minutes": None,
            "checkpoint_mb": None,
            "n_params": None,
        })
        print(f"Missing {path.name}")
        continue
    m = json.loads(path.read_text())
    test = m["test"]
    rows.append({
        "model": m["model"],
        "test_accuracy": test["accuracy"],
        "test_macro_f1": test["macro_f1"],
        "train_minutes": m["train_minutes"],
        "checkpoint_mb": m["checkpoint_mb"],
        "n_params": m["n_params"],
    })

comparison = pd.DataFrame(rows)
comparison.to_csv("comparison.csv", index=False)
comparison
```
